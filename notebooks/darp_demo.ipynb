# DARPy coverage on Python 3.14

Run `uv sync --frozen --extra coverage` from the repository root and select a
Python 3.14 notebook kernel using that project environment. Notebook frontend
and kernel tools are separate development tools. The DARPy package must be
installed in the selected kernel; no working-directory import workaround is used.

Inherited outputs and execution counts have been cleared. These migrated cells
have been syntax-checked against Python 3.14 and reviewed against current APIs;
they have not been executed as a notebook. The example is headless and limits
allocation search to 100 iterations. Numba can compile code on first use.
To request interactive graphics, install the `visualization` extra first.


```python
from darpy.coverage import MultiRobotPathPlanner
```

## Define a small rectangular coverage problem

The initial cells and obstacle use flattened grid indices.


```python
rows = 3
cols = 4
not_equal_portions = False
initial_positions = [0, 11]
portions = []
obstacles_positions = [5]
visualization = False
```

## Allocate regions and generate coverage paths

`MultiRobotPathPlanner` runs allocation and trajectory construction. Inspect
`DARP_success` before reading `best_case`; bounded search can fail.


```python
plan = MultiRobotPathPlanner(
    rows, cols, not_equal_portions, initial_positions, portions,
    obstacles_positions, visualization, MaxIter=100, seed=1,
)
if not plan.DARP_success:
    raise RuntimeError("No allocation found within the iteration budget; revise the inputs or budget")
```

```python
summary = {
    "success": plan.DARP_success,
    "iterations": plan.iterations,
    "path_edges_per_robot": [len(path) for path in plan.best_case.paths],
    "turns_per_robot": plan.best_case.turns,
}
summary
```
