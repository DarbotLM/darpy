## Compare the planner's four spanning-tree modes

This study uses the maintained rectangular-grid planner. The inherited notebook
referenced an unavailable `DARPinPoly` prototype; polygon planning is not
implemented here. The reported turns are the legacy planner's own metric,
not a proof of globally minimal turns or coordinated collision-free motion.


```python
from darpy.coverage import MultiRobotPathPlanner
```

## Define the coverage problem


```python
rows = 3
cols = 4
not_equal_portions = False
initial_positions = [0, 11]
portions = []
obstacles_positions = [5]
visualization = False
```

## Plan once and inspect each spanning-tree mode


```python
plan = MultiRobotPathPlanner(
    rows, cols, not_equal_portions, initial_positions, portions,
    obstacles_positions, visualization, MaxIter=100, seed=1,
)
if not plan.DARP_success:
    raise RuntimeError("No allocation found within the iteration budget; revise the inputs or budget")
```

```python
mode_summary = [
    {
        "mode": mode,
        "turns_per_robot": list(result.turns),
        "average_turns": float(result.avg),
        "standard_deviation": float(result.std),
    }
    for mode, result in enumerate(plan.mode_to_drone_turns)
]
mode_summary
```

## Inspect the combined result

The planner chooses the lowest reported turn count separately for each robot
across these four modes. This result can combine paths from different modes.


```python
combined_summary = {
    "turns_per_robot": list(plan.best_case.turns),
    "average_turns": float(plan.best_case.avg),
    "path_edges_per_robot": [len(path) for path in plan.best_case.paths],
}
combined_summary
```
